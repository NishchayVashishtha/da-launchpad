```python
import numpy as np
import pandas as pd

PATH = "your_file.csv"  # Set PATH to your file, for example "fintech/transactions.csv"
```

## Step 2: Load, shape and data types

```python
df = pd.read_csv(PATH)
print("rows, columns:", df.shape)
print(df.dtypes)
df.head()
```

## Steps 3 and 4: Quality, then clean

Check nulls, exact duplicates and impossible values first. Then write each fix as its own line, so a reviewer can see it.

```python
print(df.isna().sum())
print("exact duplicate rows:", df.duplicated().sum())
```

```python
# Write your cleaning steps below. Keep the cleaned table in df.
df = df.drop_duplicates().copy()
print("rows after cleaning:", len(df))
```

## Step 5: Single columns (univariate)

```python
print(df.describe(include="all").T)
for col in df.select_dtypes(exclude="number").columns[:3]:
    print(df[col].value_counts(dropna=False).head(10))
```

## Step 6: Time (trend and seasonality)

```python
DATE_COL = None  # Set DATE_COL to a date column name, for example "order_date"
if DATE_COL:
    dates = pd.to_datetime(df[DATE_COL], errors="coerce")
    print("unparsed dates:", dates.isna().sum())
    print(dates.dt.to_period("M").value_counts().sort_index())
```

## Step 7: Segments (groupby on key dimensions)

```python
SEGMENT = None  # Set SEGMENT to a text column, for example "city"
if SEGMENT:
    print(df[SEGMENT].value_counts(dropna=False).head(10))
```

## Step 8: Relationships (correlation and crosstab)

```python
num = df.select_dtypes(include="number")
print(num.corr(numeric_only=True).round(2))
```

## Step 9: Outliers and anomalies

```python
num = df.select_dtypes(include="number")
q = num.quantile([0.01, 0.5, 0.99]).T
print(q)
```
